## ⤵️ Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
import spacy
import nltk
from nltk.corpus import stopwords
```

```python
data = pd.read_csv("/content/drive/MyDrive/2026/2026 01/2 - Dados para Machine Learning: Feature Engineering/TP2/amazon_alexa.tsv", sep='\t')
data.head()
```

```python
data = data[data["verified_reviews"].notnull()]
data["verified_reviews"] = data["verified_reviews"].str.lower()
```

## 📋 Lista de exercícios

#### 📍 **Exercício 01:** Crie uma representação vetorial do tipo bag-of-words para as avaliações textuais. Imprima o número de tokens obtido, bem como uma amostra de cerca de 50 a 100 palavras do seu vocabulário.

```python
bow = CountVectorizer()

corpus = data['verified_reviews']
bow.fit_transform(corpus)

print("Vocabulary size: {}".format(len(bow.vocabulary_)))
print("Vocabulary content:\n {}".format(bow.vocabulary_))
```

```python
words = bow.get_feature_names_out()
words[::44]
```

#### 📍 **Exercício 02:** Crie uma representação vetorial do tipo bag-of-nGrams com bigramas para as avaliações textuais. Imprima o número de tokens obtido, bem como uma amostra com cerca de 50 a 100 bigramas do seu vetor.

```python
bigram = CountVectorizer(ngram_range=(2,2))
bigram.fit_transform(corpus)

print("Vocabulary size: {}".format(len(bigram.vocabulary_)))
print("Vocabulary content:\n {}".format(bigram.vocabulary_))
```

#### 📍 **Exercício 09:** Compare os resultados das questões 7 e 8 e avalie qual das técnicas – stemming ou lematização – representaria a melhor abordagem para uma análise de sentimento sobre este dataset.

```python
en_nlp = spacy.load('en_core_web_sm')
stemmer = nltk.stem.PorterStemmer()

def compare_normalization(doc):
    doc_spacy = en_nlp(doc)

    print("Stemming:")
    print([stemmer.stem(word.norm_.lower()) for word in doc_spacy])

    print("Lemmatization:")
    print([word.lemma_ for word in doc_spacy])
```

```python
for i in range(100, 105):
  compare_normalization(corpus[i])
  print("\n")
```

🗣️ Imagino que, para uma análise de sentimentos, o melhor seria a utilização do método Lemmatization, pois as palavras ficam mais claras (sinônimos). Caso utilizassemos o Stemming, teríamos algumas palavras incompletas (apenas a raíz da mesma).
